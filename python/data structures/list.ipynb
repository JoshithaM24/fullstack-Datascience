# 📋 Python Lists

# A list is a built-in, mutable data structure used to store
# an ordered collection of items.

# Creating a List
my_list = [1, 2, 3]

print(my_list)


# 🔹 Characteristics of Lists

# 1. Ordered → Items maintain insertion order
# 2. Mutable → Elements can be added, removed, or modified
# 3. Allows Duplicates → Same value can appear multiple times
# 4. Index-based → Index starts from 0
# 5. Heterogeneous → Can store different data types together


| **Task**     | **Method / Function** | **Simple Meaning**                        |
| ------------ | --------------------- | ----------------------------------------- |
| **Add**      | `append()`            | Adds **one item** at the end              |
| **Add**      | `extend()`            | Adds **multiple items**                   |
| **Add**      | `insert()`            | Adds item at a **specific index**         |
| **Remove**   | `remove()`            | Removes item **by value**                 |
| **Remove**   | `pop()`               | Removes item **by index**                 |
| **Remove**   | `clear()`             | Removes **all items**                     |
| **Organize** | `sort()`              | Sorts the **original list**               |
| **Organize** | `reverse()`           | Reverses the **original list**            |
| **Organize** | `sorted()`            | Returns a **new sorted list**             |
| **Info**     | `len()`               | Returns **number of items**               |
| **Info**     | `count()`             | Counts how many times a value occurs      |
| **Info**     | `index()`             | Returns the **index/position** of a value |
| **Info**     | `sum()`               | Adds **numeric values**                   |
| **Creation** | `list()`              | Creates/converts to a list                |
| **Creation** | `copy()`              | Creates a **shallow copy**                |


| **Method**  | **Syntax**                 | **Example**        |
| ----------- | -------------------------- | ------------------ |
| `append()`  | `list.append(item)`        | `a.append(4)`      |
| `extend()`  | `list.extend(iterable)`    | `a.extend([4, 5])` |
| `insert()`  | `list.insert(index, item)` | `a.insert(1, 10)`  |
| `remove()`  | `list.remove(item)`        | `a.remove(2)`      |
| `pop()`     | `list.pop(index)`          | `a.pop(1)`         |
| `clear()`   | `list.clear()`             | `a.clear()`        |
| `index()`   | `list.index(item)`         | `a.index(20)`      |
| `count()`   | `list.count(item)`         | `a.count(2)`       |
| `sort()`    | `list.sort()`              | `a.sort()`         |
| `reverse()` | `list.reverse()`           | `a.reverse()`      |
| `sorted()`  | `sorted(list)`             | `sorted(a)`        |
| `len()`     | `len(list)`                | `len(a)`           |
| `sum()`     | `sum(list)`                | `sum(a)`           |
| `list()`    | `list(iterable)`           | `list((1,2,3))`    |
| `copy()`    | `list.copy()`              | `b = a.copy()`     |


| **Characteristic**    | **Meaning**                          |
| --------------------- | ------------------------------------ |
| **Ordered**           | Items maintain their order           |
| **Mutable**           | Items can be changed                 |
| **Allows Duplicates** | Same value can appear multiple times |
| **Index-based**       | Index starts from `0`                |
| **Heterogeneous**     | Can store different data types       |


Easy revision:
append = one → extend = many → insert = position → remove = value → pop = index → sort = original → sorted = new list
    

# 📚 List Methods Reference

# ADD
# append()  → Adds one item
# extend()  → Adds multiple items
# insert()  → Adds item at a specific index

# REMOVE
# remove()  → Removes an item by value
# pop()     → Removes an item by index
# clear()   → Removes all items

# ORGANIZE
# sort()    → Sorts the original list
# reverse() → Reverses the original list
# sorted()  → Returns a new sorted list

# INFORMATION
# len()     → Returns number of items
# count()   → Counts occurrences
# index()   → Returns index of an item
# sum()     → Adds numeric values

# CREATION
# list()    → Creates a list
# copy()    → Creates a shallow copy
