📘 Features of Python Dictionary

A Python dictionary is a powerful and flexible data structure used to store data in key-value pairs.

| **Feature**                 | **Description**                                                |
| --------------------------- | -------------------------------------------------------------- |
| **Key-Value Structure**     | Data is stored as **key : value** pairs                        |
| **Mutable**                 | Items can be **added, updated, or removed**                    |
| **Ordered**                 | Maintains **insertion order** (Python 3.7+)                    |
| **Unique Keys**             | Duplicate keys are **not allowed**                             |
| **Immutable Keys**          | Keys must be **hashable**, such as strings, numbers, or tuples |
| **Allows Duplicate Values** | Multiple keys can have the **same value**                      |
| **Fast Access**             | Values can be retrieved quickly using **keys**                 |
| **Dynamic Size**            | Dictionary can **grow or shrink** dynamically                  |


format : d = (Key:Value)

```python
student = {
    "name": "Joshitha",
    "age": 21,
    "city": "Hyderabad"
}
print(student)
```

| **Method** | **Purpose**                                     |
| ---------- | ----------------------------------------------- |
| `keys()`   | Returns all **keys**                            |
| `values()` | Returns all **values**                          |
| `items()`  | Returns all **key-value pairs**                 |
| `get()`    | Gets a value **safely using a key**             |
| `update()` | **Adds or updates** key-value pairs             |
| `pop()`    | **Removes and returns** a specified key's value |
| `clear()`  | **Removes all items**                           |


```python
person = {
    "name": "Joshitha",
    "age": 21,
    "city": "Hyderabad"
}
print(person)
```

```python
print(f"Person Name {person['name']}")
print(f"Person Age {person['age']}")
print(f"Person course {person['city']}")
```

```python

print(f"Keys : {person.keys()}")
print(f"Values : {person.values()}")
```

```python
# Pop is based on LIFO
person = {
    "name": "joah",
    "age": 33,
    "course": "Python",
    "guest": "spiderman",
    "special": "nycc",
}
person.popitem()
```

```python
# Update the dictionary
person.update(city="Hyderabad")
person.update({"data": "Python"})
print(f"Updated Data: {person}")
```

```python
# Update the same keys dictionary
person.update(city="Hyderabad1")
person.update({"data": "Python1"})
print(f"Updated Data: {person}")
```

```python
person['data']='Test'
print(f"Updated Data: {person}")
```

```python
# Adding the new keys dictionary
person['test']='Python'
print(f"Updated Data: {person}")
```

```python
data=person.copy()
print(f"Copied Data {data}")
```

```python
# Adding new item with default None value
data.setdefault('new')  #setdefault() checks whether a key already exists in the dictionary.
print(f"Copied Data {data}")
```

```python
data.clear()
print(f"Cleared Data {data}")
```
