# 🔢 Python Range Data Structure

# range() is an immutable sequence used to generate
# a series of numbers.

# It is commonly used for:
# ✔ Loops
# ✔ Iteration
# ✔ Index generation

| **Feature**           | **Description**                                 |
| --------------------- | ----------------------------------------------- |
| **Immutable**         | Values cannot be changed                        |
| **Ordered**           | Numbers follow a specific sequence              |
| **Memory Efficient**  | Does not store all numbers in memory            |
| **Integer Sequence**  | Generates a sequence of integers                |
| **Supports Indexing** | Individual values can be accessed using indexes |



| **Syntax**                 | **Meaning**                             | **Example**                 |
| -------------------------- | --------------------------------------- | --------------------------- |
| `range(stop)`              | Starts from `0`, ends before `stop`     | `range(5)` → `0,1,2,3,4`    |
| `range(start, stop)`       | Starts from `start`, ends before `stop` | `range(2,6)` → `2,3,4,5`    |
| `range(start, stop, step)` | Moves by `step`                         | `range(2,10,2)` → `2,4,6,8` |


```python
range(5)
```

```python
list(range(5))
```

```python

list(range(0,5,2))
```

```python
data_range = list(range(0,15,3))
print(data_range)
```

| **Code**             | **Result**        |
| -------------------- | ----------------- |
| `range(1, 6)`        | `range(1, 6)`     |
| `tuple(range(1, 6))` | `(1, 2, 3, 4, 5)` |
| `set(range(1, 6))`   | `{1, 2, 3, 4, 5}` |
| `list(range(1, 6))`  | `[1, 2, 3, 4, 5]` |


```python
data_range[1]
```

```python
# List operations can be performed since it's in the list
data_range[1] =4
print(data_range)
```
