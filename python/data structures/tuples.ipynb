📦 Python Tuples

A tuple is a built-in data structure used to store an ordered collection of items.

Tuples are similar to lists, but they are immutable, meaning their values cannot be changed after creation.

| **Characteristic**    | **Description**                                  |
| --------------------- | ------------------------------------------------ |
| **Immutable**         | Tuple items **cannot be changed** after creation |
| **Ordered**           | Items maintain their **insertion order**         |
| **Allows Duplicates** | Duplicate values are allowed                     |
| **Index-based**       | Elements are accessed using **indexes**          |
| **Heterogeneous**     | Can store **different data types**               |



| **Task**     | **Method / Function** | **Simple Meaning**                    |
| ------------ | --------------------- | ------------------------------------- |
| **Count**    | `count()`             | Counts how many times a value appears |
| **Find**     | `index()`             | Returns the index of a value          |
| **Info**     | `len()`               | Returns the number of items           |
| **Info**     | `sum()`               | Adds numeric values                   |
| **Creation** | `tuple()`             | Creates/converts to a tuple           |



**Method / Function** | **Syntax**          | **Example**        |
| --------------------- | ------------------- | ------------------ |
| `count()`             | `tuple.count(item)` | `t.count(2)`       |
| `index()`             | `tuple.index(item)` | `t.index(2)`       |
| `len()`               | `len(tuple)`        | `len(t)`           |
| `sum()`               | `sum(tuple)`        | `sum(t)`           |
| `tuple()`             | `tuple(iterable)`   | `tuple([1, 2, 3])` |




| **Code**          | **Result**             |
| ----------------- | ---------------------- |
| `my_tuple = (1,)` | ✅ Tuple                |
| `my_tuple = (1)`  | ❌ Integer, not a tuple |

```python
# Mixed data types
t1 = (1,2,3,45.6,67)
t2 = ('jo',1,2,45.5,56,1+2j,True)
print(type(t1))
print(type(t2))
```

```python
# Single-Element Tuples Need a Comma
t3 = (4)
print(type(t3))
t4 = (4,)
print(type(t4))
```

```python
# Count() returns number of occurances
t5 = (10,'jo',45,35,10,23,10,10.45)
print(t5.count(10))

# Get the index of value
print(t5.index(10))
```

```python

# Tuple can't be mutable - Immutable
print(t5[2])
t5[2] = 10
```

```python
# We can modify mutable objects inside tuple like tuple consists list we can modify. But tuple objects we can't modify
tup = (['foo',[1,2],True])
tup[1][0]=3
print(tup)
```

Unpacking tuples


```python
tup = (3,4,5)
a,b,c = tup
print(a)
print(b)
print(c)
```

```python
# number of variable should be equal of values
a,b = tup
print(a)
print(b)
```
