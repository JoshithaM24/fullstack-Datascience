### 1. Arithmetic Operators

Used for mathematical calculations.

| **Operator** | **Name** | **Example** | **Result** |
|---|---|---|---:|
| `+` | Addition | `10 + 5` | 15 |
| `-` | Subtraction | `10 - 5` | 5 |
| `*` | Multiplication | `10 * 5` | 50 |
| `/` | Division | `10 / 4` | 2.5 |
| `//` | Floor Division | `10 // 4` | 2 |
| `%` | Modulus | `10 % 3` | 1 |
| `**` | Exponentiation | `2 ** 3` | 8 |

```python
x = 5 
y = 10
```

```python
x + y
```

```python
x - y
```

```python
x * y
```

```python
x / y
```

```python
x // y
```

```python
y % x
```

```python
y ** x
```

### 2. Assignment Operators

Assignment operators are used to assign values to variables.

Python also supports **compound assignment operators**, which perform an operation and assignment in a single step.

---

| **Operator** | **Example** | **Equivalent To** |
|---|---|---|
| `=` | `x = 5` | Assigns value 5 to x |
| `+=` | `x += 3` | `x = x + 3` |
| `-=` | `x -= 2` | `x = x - 2` |
| `*=` | `x *= 4` | `x = x * 4` |
| `/=` | `x /= 2` | `x = x / 2` |
| `//=` | `x //= 2` | `x = x // 2` |
| `%=` | `x %= 3` | `x = x % 3` |
| `**=` | `x **= 2` | `x = x ** 2` |

```python
x = 10
```

```python
x += 5
print(x)
```

```python

x *= 2
print(x)
```

```python
x -= 10
print(x)  
```

### 3. Comparison (Relational) Operators

These operators compare two values and return a Boolean value (`True` or `False`).

| **Operator** | **Meaning** |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

```python
x = 5
y = 10
z = 5
```

```python
x == y
```

```python
x != y
```

```python
x == z
```

```python
x > y
```

```python
x >= y
```

```python
x <= z
```

### 4. Logical Operators

Used to combine conditional statements.

| **Operator** | **Description** |
|---|---|
| **and** | Returns `True` if both conditions are true |
| **or** | Returns `True` if at least one condition is true |
| **not** | Reverses the result |

```python
x = 7

print(x > 5 and x < 10)   
print(x < 5 or x > 10)    
print(not(x > 5))         
```

```python
not ( x < 10 )
```

```python
x < y and z < y
```

```python
x < y or z > y
```

### 5. Unary Operator

Unary operators work with **only one operand**.

Python supports the following primary unary operator:

---

#### 🔹 Unary Minus (`-`)

The unary minus operator is used to **negate** a numeric value.

It changes:

- Positive → Negative
- Negative → Positive

```python
x = 5
print(-x)
```

```python
x = 10
```

```python
-x
```

```python
x = -20
-x
```
