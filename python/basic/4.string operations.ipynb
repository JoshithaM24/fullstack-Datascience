
String Operations

CORE STRING OPERATI0NS

| Operator                    | Description                                     | Example               |
| --------------------------- | ----------------------------------------------- | --------------------- |
| **Concatenation (+)**       | Joins two or more strings                       | `"Hello" + "World"`   |
| **Repetition (*)**          | Repeats a string multiple times                 | `"Hi" * 3 → "HiHiHi"` |
| **Membership (in, not in)** | Checks if substring exists (returns True/False) | `"H" in "Hello"`      |
| **Indexing ([])**           | Access a character by position (0-based)        | `text[0]`             |
| **Slicing ([:])**           | Extract part of a string using range            | `text[1:4]`           |


### Essential Built-in String Methods

These methods are called directly on a string object (e.g., `text.upper()`)

| **Category** | **Common Methods** | **Description** |
|---|---|---|
| **Case Changes** | `upper(), lower(), capitalize(), title()` | Converts the string to uppercase, lowercase, or capitalized formats |
| **Search & Find** | `find(), index(), count()` | Locates a substring's position or counts its occurrences |
| **Cleaning** | `strip(), lstrip(), rstrip()` | Removes leading and trailing whitespace or specific characters |
| **Split & Join** | `split(), join()` | `split()` breaks a string into a list; `join()` merges a list into one string |
| **Replacement** | `replace()` | Substitutes a specified substring with another |
| **Validation** | `isdigit(), isalpha(), isalnum()` | Returns True if the string contains only digits, letters, or both |

1. Indexing
   
➡️ Forward Indexing (Positive)
Starts from 0 for the first character

Moves left → right

Start: Index 0 → first character

End: Index length - 1 → last character

⬅️ Backward Indexing (Negative)
Starts from -1 for the last character

Moves right → left

Start: Index -1 → last character

End: Index -length → first character



## Visual Comparison

Using the word **"Python"** as an example

| **Character** | **P** | **y** | **t** | **h** | **o** | **n** |
|---|---|---|---|---|---|---|
| **Forward Index** | 0 | 1 | 2 | 3 | 4 | 5 |
| **Backward Index** | -6 | -5 | -4 | -3 | -2 | -1 |

## 2. Slicing

The most common way to slice a string is using bracket notation: `string[start:stop:step]`

| **Parameter** | **Explanation** |
|---|---|
| **start** | The index where the slice begins (**inclusive**). Default = 0 |
| **stop** | The index where the slice ends (**exclusive**) |
| **step** | (Optional) Interval between characters. Default = 1 |

```python
data_python = "welcome to python programming"
```

```python
data_python[:]           # Return all characters
```

```python
data_python[1:6]         # Return 1 to 5 characters
```

```python
data_python[8:]          # Return 8 to end characters
```

```python
data_python[:10]         # Return start to 9 characters
```

```python
data_python[:5]          # Return start to 4 characters
```

```python
data_python[-1:]         # Return last character
```

```python
data_python[-6:]         # Return last 6th character to end
```

```python
data_python[-7:-2]       # Return last 7th character to 3rd characterd from last
```

```python
data_python[1000:-2]     # Returns empty since so index of 1000
```

```python

data_python[10000:]     # Returns empty since so index of 10000
```

```python
data_python[0:1000]     # Returns all since stop index till 1000
```

```python
data_python[0:10:2]     # Returns step slicing returns 0,2,4,6,8
```

```python
data_python[5::2]       # Returns step slicing returns 5,7,9,... to end
```
