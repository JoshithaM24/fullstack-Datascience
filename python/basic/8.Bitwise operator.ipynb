## Bitwise Operators in Python

Bitwise operators manipulate integer data at the binary level (individual bits: 0s and 1s). They are commonly used in low-level programming, embedded systems, networking, and efficient data processing.

| **Operator** | **Name** | **Description** |
|---|---|---|
| `&` | AND | Sets a bit to `1` if both corresponding bits are `1`. |
| `\|` | OR | Sets a bit to `1` if at least one corresponding bit is `1`. |
| `^` | XOR | Sets a bit to `1` if the corresponding bits are different. |
| `~` | NOT | Inverts the bits. In Python, `~n` is equivalent to `-(n + 1)`. |
| `<<` | Left Shift | Shifts bits to the left; generally equivalent to multiplying by powers of 2. |
| `>>` | Right Shift | Shifts bits to the right; generally equivalent to floor division by powers of 2. |
