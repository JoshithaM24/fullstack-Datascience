## 5. `input()` Function

In Python, the `input()` function is a built-in function that allows a program to receive data from the user through the keyboard. When called, it pauses the program and waits for the user to enter something and press **Enter**.

### Key Features

#### 1. Returns a String

Regardless of what the user enters (even if it is a number), the `input()` function always returns the data as a string (`str`).

#### 2. Optional Prompt

You can provide a message to guide the user, such as `"Enter your name: "`, or leave it empty to simply wait for input.

#### 3. Pauses Execution

The program will not proceed to the next line until the user enters the data and presses the **Enter** key.

```python
x = input("Enter the 1st string")
y = input("Enter the 2nd string")
z = x + ' ' + y
print (z)
```

```python

a = int(input("Enter the 1st number"))
b = int(input("Enter the 2nd number"))
c = a + b
print (f"Total sum of 2 numbers {c}")
```

## 6. `eval()` Function
The `eval()` function executes a Python expression provided as a string and returns the result.
eval(expression, globals=None, locals=None)



| **Parameter** | **Description** |
|---|---|
| **expression** | A string containing the Python expression to evaluate |
| **globals** | (Optional) Dictionary for global variables |
| **locals** | (Optional) Dictionary for local variables |

```python

result = eval(input("Enter expression to eval"))
print(f"Expression result {result}")
```

```python
    X = eval(input("Enter expression to eval"))
print(f"Expression result {X}")
```
