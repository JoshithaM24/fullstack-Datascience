## Number System in Python

Python supports four primary number systems for representing integers: Decimal, Binary, Octal, and Hexadecimal. While these are used to represent values in different bases, they all ultimately belong to the `int` data type in Python.

| **Number System** | **Base** | **Digits Used** | **Prefix** | **Example** |
|---|---:|---|---|---|
| **Decimal** | 10 | 0–9 | No prefix | `10` |
| **Binary** | 2 | 0, 1 | `0b` or `0B` | `0b1010` |
| **Octal** | 8 | 0–7 | `0o` or `0O` | `0o12` |
| **Hexadecimal** | 16 | 0–9, A–F | `0x` or `0X` | `0xA` |
