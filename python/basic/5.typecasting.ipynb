## Type Casting

Convert one data type to another data type.

### 1. Convert All Data Types to `int`

| **Rule** | **Explanation** |
|---|---|
| **Single Argument** | In type casting to `int`, we can pass only **one argument**. |
| **Complex → int** | We **cannot** type cast a complex number to `int`. |
| **String → int** | We **cannot** type cast a text string to `int`. |
| **Other Data Types → int** | We can type cast the remaining compatible data types to `int`. |

```python
int (3.5)   #float to int type cast
```

```python
int (True)     # Boolean to int type cast
```

```python
int ('10')     # Number String to int type cast
```

```python
int (False)     # Boolean to int type cast
```

```python
int (3 + 5j)   # Can't convert complex to int
```

```python
int ('10.5')   # Can't convert Text string to int type cast
```

```python
int ('ten')   # Can't convert Text string to int type cast
```

```python
int (10.5,23.5) # can't pass 2 arguments
```

## 2. Convert All Data Types to `float`

| **Rule** | **Explanation** |
|---|---|
| **Single Argument** | In type casting to `float`, we can pass only **one argument**. |
| **Complex → float** | We **cannot** type cast a complex number to `float`. |
| **String → float** | We **cannot** type cast a text string to `float`. |
| **Other Data Types → float** | We can type cast the remaining compatible data types to `float`. |

```python
float ( 23 )      # Convert type to float
```

```python
float ( True )    # Convert boolean to float
```

```python
float ( 3 + 5j )  # Can't convert complex to float
```

```python
float ( 2, 3 )    # can't pass 2 parameters to float type
```

```python
float ('23')      # Convert Number string to float
```

```python
float ('23.5')    # Convert Text string to float
```

```python
float ('ten')     # Convert Text string to float
```

## 3. Convert All Data Types to `bool`

| **Rule** | **Explanation** |
|---|---|

```python
bool (2.5)        # convert float to boolean
```

```python
bool (2.5)        # convert float to boolean
```

```python
bool (None)       # convert none to boolean
```

## 4. Convert All Data Types to `complex`

| **Rule** | **Explanation** |
|---|---|
| **Maximum Arguments** | In type casting to `complex`, we cannot pass **3 arguments or more**. |
| **Second Argument** | We cannot pass the **second argument as a string**. |
| **Other Data Types → complex** | We can type cast the remaining compatible data types to `complex`. |

```python
complex (10)       # convert int to complex
```

```python
complex (4.5)      # convert float to complex
```

```python
complex (10,4.5)   # convert 2 params to complex
```

```python
complex (10,4.5,6) # can't pass 3 parameters
```

```python
complex (10,20)    # convert int to complex
```

```python
complex (2.3,4.5)  # convert float to complex
```

```python
complex (True)     # convert boolean to complex
```

```python
complex (False)    # convert boolean to complex
```

```python
complex (True,True)   # convert boolean to complex
```

```python
complex (True,False)  # convert boolean to complex
```

```python
complex (False,False) # convert boolean to complex
```

```python
complex (False,True)  # convert boolean to complex
```

```python
complex ('10')    # convert number string to complex
```

```python
complex ('10','20')    # can't convert 2 args number string to complexcomplex ('10','20')    
```

```python
complex (10,'20')      # can't convert 2 args number string to complex
```

## 5. Convert All Data Types to `str`

| **Rule** | **Explanation** |
|---|---|
| **Maximum Arguments** | In type casting to `str`, we cannot pass **2 arguments or more**. |
| **All Data Types → str** | We can type cast **all data types** to `str`. |

```python
str (20)    # convert int to string
```

```python
str (23.4)  # convert float to string
```

```python
str (True)  # convert boolean to string
```

```python
str (3+4j)  # convert complex to string
```
