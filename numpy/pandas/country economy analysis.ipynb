### Pandas – Quick Revision Table

| Command | Use |
|---|---|
| `pd.read_excel(path)` | Read Excel file |
| `id(df)` | Get object identity |
| `pd.__version__` | Check Pandas version |
| `len(df)` | Count rows |
| `df.shape` | Get rows and columns |
| `df.head()` | First 5 rows |
| `df.head(n)` | First n rows |
| `df.tail()` | Last 5 rows |
| `df.columns` | Display column names |
| `df.info()` | Check data types and non-null counts |
| `df.describe()` | Statistical summary |
| `df['CountryName']` | Select one column |
| `df[['CountryName', 'BirthRate']]` | Select multiple columns |
| `df.iloc[0]` | Select first row by position |
| `df.isnull().sum()` | Count missing values per column |

**Remember:**
- `len(df)` → rows only
- `df.shape` → `(rows, columns)`
- `head()` → first 5 rows by default
- `dtypes&columns`
- `axix-0 rows | axix-1 columns|`
|`pandas if values are |number --int|decimal--float|text--object`|
